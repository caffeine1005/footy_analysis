```python
import polars as pl
import matplotlib.pyplot as plt
```

```python
all_player = pl.read_csv("player_stats.csv")
all_player
```

```python
all_player = all_player = all_player.filter((pl.col("('pos', '')").str.contains("MF")))
all_player_copy = all_player
```

```python
dribble_succ = all_player["player","('Touches', 'Touches')"]
```

```python
dribble_succ = dribble_succ.drop_nulls()
```

```python
dribble = all_player["player","('Progression', 'PrgP')"].drop_nulls()
```

```python
dribbles = dribble.join(dribble_succ, on="player", how="inner")
```

```python
dribbles
```

```python
from scipy.stats import pearsonr
import numpy as np

x = all_player_copy["('SCA', 'SCA')"]
y = all_player_copy["('Expected', 'xAG')"]

x = np.asarray(x.to_list())
y = np.asarray(y.to_list())

corr_coefficient, p_value = pearsonr(x, y)
```

```python
corr_coefficient
```

```python
fig, ax = plt.subplots()

ax.scatter(dribbles["('Touches', 'Touches')"], dribbles["('Progression', 'PrgP')"])
    
plt.xlabel("('Take-Ons', 'Succ')")
plt.ylabel("('Take-Ons', 'Att')")
```
